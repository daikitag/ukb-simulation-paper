# LD Decay Plot

This Jupyter Notebook produces an LD decay plot for chromosome 1p. The data is obtained by using `obtain_ld_decay_summary.py`.

```python
import pickle
import pandas as pd
import matplotlib.pyplot as plt
```

```python
with open('thousand_ld_decay_plot_chr1p.pcl', 'rb') as f:
    thousand_data = pickle.load(f)
```

```python
with open('simulation_ld_decay_plot_chr1p.pcl', 'rb') as f:
    simulation_data = pickle.load(f)
```

```python
plt.rcParams.update({'font.size': 18})

i = 1
fig = plt.figure(figsize=(14, 10))
for pop in ["ceu", "yri", "chb", "jpt"]:
    plt.subplot(2, 2, i)
    plt.plot(
        thousand_data[pop]["x_bp"] / 1000,
        thousand_data[pop]["mean_r2"],
        lw=2,
        label="1000 Genomes Project",
    )
    plt.plot(
        simulation_data[pop]["x_bp"] / 1000,
        simulation_data[pop]["mean_r2"],
        lw=2,
        label="Simulation",
    )
    if i == 1:
        plt.legend()
    plt.title(f"Population: {pop.upper()}", fontsize=25)
    #plt.xlabel("Distance (Kb)")
    #plt.ylabel("Mean r2")
    i += 1

plt.subplots_adjust(hspace=0.3, wspace=0.2)
fig.supxlabel("Distance (kb)", fontsize=25)
fig.supylabel(r"Mean $r^2$", fontsize=25)
fig.suptitle(f"LD Decay Plot: Chromosome 1q", fontsize=30, y=1)
plt.savefig("ld-decay-chr1q.pdf")
```
